```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, pauli_error

# Creating 3 qubits, 3 classical bits to record the collapased states of the qubits upon measurement
qc = QuantumCircuit(3, 3)

# Bell pair Preparation: |Phi+> = 1/sqrt(2) * (|00> + |11>)
# Putting qubit-0 in superposition
qc.h(0)

# Applying a CNOT gate with qubit 0 as control and qubit 1 as target - linking both qubits
qc.cx(0, 1)

qc.barrier()

# Error check
qc.cx(0, 2)
qc.cx(1, 2)

# 0,1,2 qubits measurement into classical bits 0,1,2
qc.measure([0, 1, 2], [0, 1, 2])

# Noise Model (10% bit-flip probability)
noise_model = NoiseModel()
error = pauli_error([('X', 0.1), ('I', 0.9)])
noise_model.add_all_qubit_quantum_error(error, ['h'])
noise_model.add_all_qubit_quantum_error(error.tensor(error), ['cx'])

# 3. Executions
shots = 1000
sim_ideal = AerSimulator()
sim_noisy = AerSimulator(noise_model=noise_model)

raw_ideal = sim_ideal.run(qc, shots=shots).result().get_counts()
raw_noisy = sim_noisy.run(qc, shots=shots).result().get_counts()

# Helper to slice 3-bit output strings down to 2-bit data outcomes ('c1 c0')
def extract_2bit_counts(raw_counts):
    counts_2bit = {}
    for bitstring, count in raw_counts.items():
        data_bits = bitstring[1:]  # Keep only data qubits c1 c0
        counts_2bit[data_bits] = counts_2bit.get(data_bits, 0) + count
    return counts_2bit

# Map outputs to 2 bits
ideal_counts = extract_2bit_counts(raw_ideal)
noisy_counts = extract_2bit_counts(raw_noisy)

# Protected Case: Keep 2-bit state only if ancilla bit c2 == '0'
clean_protected_counts = {}
for bitstring, count in raw_noisy.items():
    ancilla_bit = bitstring[0]  # c2
    data_bits = bitstring[1:]   # c1 c0
    if ancilla_bit == '0':
        clean_protected_counts[data_bits] = clean_protected_counts.get(data_bits, 0) + count

# 5. Output
print("1. Ideal Bell Pair Counts:", ideal_counts)
print("2. Noisy Bell Pair Counts:", noisy_counts)
print("3. Protected Bell Pair Counts:", clean_protected_counts)

print("\n1. Ideal Bell Pair Fidelity:", (ideal_counts['00']+ideal_counts['11'])/shots)
print("2. Noisy Bell Pair Fidelity:", round((noisy_counts['00']+noisy_counts['11'])/shots,2))
print("3. Protected Bell Pair Fidelity:", round((clean_protected_counts['00']+clean_protected_counts['11'])/(clean_protected_counts['00']+clean_protected_counts['11']+clean_protected_counts['01']+clean_protected_counts['10']),2))
```
